```python
# ==============================
# PRACTICAL WORK 2
# Student and Course Management
# ==============================

students = []
courses = []
marks = []


# ------------------------------
# Input students
# ------------------------------

number_of_students = int(input("Enter number of students: "))

for i in range(number_of_students):
    student_id = input("Student ID: ")
    student_name = input("Student name: ")
    dob = input("Date of birth: ")

    student = [student_id, student_name, dob]
    students.append(student)


# ------------------------------
# Input courses
# ------------------------------

number_of_courses = int(input("\nEnter number of courses: "))

for i in range(number_of_courses):
    course_id = input("Course ID: ")
    course_name = input("Course name: ")

    course = [course_id, course_name]
    courses.append(course)


# ------------------------------
# Input marks
# ------------------------------

for student in students:
    for course in courses:
        mark = float(input(
            "Enter mark for " + student[1] + " - " + course[1] + ": "
        ))

        mark_information = [
            student[0],
            course[0],
            mark
        ]

        marks.append(mark_information)


# ------------------------------
# Functions
# ------------------------------

def list_students():
    print("\n===== STUDENTS =====")

    for student in students:
        print(
            "ID:", student[0],
            "| Name:", student[1],
            "| DOB:", student[2]
        )


def list_courses():
    print("\n===== COURSES =====")

    for course in courses:
        print(
            "ID:", course[0],
            "| Name:", course[1]
        )


def show_marks():
    print("\n===== MARKS =====")

    for mark in marks:
        print(
            "Student:", mark[0],
            "| Course:", mark[1],
            "| Mark:", mark[2]
        )


# ------------------------------
# Display information
# ------------------------------

list_students()
list_courses()
show_marks()s
```

```python
#step2
# ==============================
# PRACTICAL WORK 3 - STEP 2
# Math module and floor()
# ==============================

import math

students = []
courses = []
marks = []


# ------------------------------
# Input students
# ------------------------------

number_of_students = int(input("Enter number of students: "))

for i in range(number_of_students):
    student_id = input("Student ID: ")
    student_name = input("Student name: ")
    dob = input("Date of birth: ")

    student = [student_id, student_name, dob]
    students.append(student)


# ------------------------------
# Input courses
# ------------------------------

number_of_courses = int(input("\nEnter number of courses: "))

for i in range(number_of_courses):
    course_id = input("Course ID: ")
    course_name = input("Course name: ")

    course = [course_id, course_name]
    courses.append(course)


# ------------------------------
# Input marks
# ------------------------------

for student in students:
    for course in courses:

        mark = float(input(
            "Enter mark for " + student[1] + " - " + course[1] + ": "
        ))

        # Round DOWN to 1 decimal place
        mark = math.floor(mark * 10) / 10

        mark_information = [
            student[0],
            course[0],
            mark
        ]

        marks.append(mark_information)


# ------------------------------
# Functions
# ------------------------------

def list_students():
    print("\n===== STUDENTS =====")

    for student in students:
        print(
            "ID:", student[0],
            "| Name:", student[1],
            "| DOB:", student[2]
        )


def list_courses():
    print("\n===== COURSES =====")

    for course in courses:
        print(
            "ID:", course[0],
            "| Name:", course[1]
        )


def show_marks():
    print("\n===== MARKS =====")

    for mark in marks:
        print(
            "Student:", mark[0],
            "| Course:", mark[1],
            "| Mark:", mark[2]
        )


# ------------------------------
# Display information
# ------------------------------

list_students()
list_courses()
show_marks()
```
