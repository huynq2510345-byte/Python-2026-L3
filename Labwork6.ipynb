```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5, 5, 100)
y = 2*x + np.sin(x)**2 + 3
plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title("hello, po")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("scores.csv")

# 1. Scatter plot
plt.scatter(df["math"], df["python"], label="Math vs Python")
plt.xlabel("Math")
plt.ylabel("Python")
plt.savefig("scattered.png")
plt.show()

# 2. Histograms: three separate subplots
df[["math", "python", "db"]].hist()
plt.savefig("histo.png")
plt.show()

# 3. Box plot
df[["math", "python", "db"]].plot(kind="box")
plt.savefig("box.png")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "math": [70, 80, 90, 60, 85],
    "python": [75, 85, 95, 65, 80],
    "db": [80, 75, 90, 70, 85]
})

plt.scatter(df["math"], df["python"])
plt.xlabel("Math")
plt.ylabel("Python")
plt.savefig("scattered.png")
plt.show()

df[["math", "python", "db"]].hist()
plt.savefig("histo.png")
plt.show()

df[["math", "python", "db"]].plot(kind="box")
plt.savefig("box.png")
plt.show()
```
