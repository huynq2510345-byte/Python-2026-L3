```python
# ==============================
# PRACTICAL WORK 3 - STEP 3
# NumPy + Weighted GPA
# ==============================

import math
import numpy as np

students = []
courses = []
marks = []


# ------------------------------
# Input students
# ------------------------------

number_of_students = int(input("Enter number of students: "))

for i in range(number_of_students):
    student_id = input("Student ID: ")
    student_name = input("Student name: ")
    dob = input("Date of birth: ")

    student = [student_id, student_name, dob]
    students.append(student)


# ------------------------------
# Input courses
# ------------------------------

number_of_courses = int(input("\nEnter number of courses: "))

for i in range(number_of_courses):
    course_id = input("Course ID: ")
    course_name = input("Course name: ")
    credit = int(input("Number of credits: "))

    course = [course_id, course_name, credit]
    courses.append(course)


# ------------------------------
# Input marks
# ------------------------------

for student in students:
    for course in courses:

        mark = float(input(
            "Enter mark for " + student[1] + " - " + course[1] + ": "
        ))

        # Round DOWN to 1 decimal place
        mark = math.floor(mark * 10) / 10

        mark_information = [
            student[0],
            course[0],
            mark
        ]

        marks.append(mark_information)


# ------------------------------
# Calculate GPA
# ------------------------------

def calculate_gpa(student_id):

    weighted_marks = []
    credits = []

    for mark in marks:

        if mark[0] == student_id:

            course_id = mark[1]
            student_mark = mark[2]

            for course in courses:

                if course[0] == course_id:

                    credit = course[2]

                    weighted_marks.append(student_mark * credit)
                    credits.append(credit)

    weighted_marks = np.array(weighted_marks)
    credits = np.array(credits)

    if np.sum(credits) == 0:
        return 0

    gpa = np.sum(weighted_marks) / np.sum(credits)

    return gpa


# ------------------------------
# Functions
# ------------------------------

def list_students():

    print("\n===== STUDENTS =====")

    for student in students:

        gpa = calculate_gpa(student[0])

        print(
            "ID:", student[0],
            "| Name:", student[1],
            "| DOB:", student[2],
            "| GPA:", round(gpa, 2)
        )


def list_courses():

    print("\n===== COURSES =====")

    for course in courses:

        print(
            "ID:", course[0],
            "| Name:", course[1],
            "| Credits:", course[2]
        )


def show_marks():

    print("\n===== MARKS =====")

    for mark in marks:

        print(
            "Student:", mark[0],
            "| Course:", mark[1],
            "| Mark:", mark[2]
        )


# ------------------------------
# Display information
# ------------------------------

list_students()
list_courses()
show_marks()
```

```python
# ------------------------------
# Sort students by GPA
# ------------------------------

def sort_students_by_gpa():

    student_gpa = []

    for student in students:

        gpa = calculate_gpa(student[0])

        student_gpa.append([student, gpa])

    # Sort from highest GPA to lowest GPA
    student_gpa.sort(key=lambda x: x[1], reverse=True)

    print("\n===== STUDENTS SORTED BY GPA =====")

    for item in student_gpa:

        student = item[0]
        gpa = item[1]

        print(
            "ID:", student[0],
            "| Name:", student[1],
            "| GPA:", round(gpa, 2)
        )


# ------------------------------
# Display sorted students
# ------------------------------

sort_students_by_gpa()
```

```python
import math
import numpy as np
import courses


students = []
course_list = []
marks = []


# ------------------------------
# Input students
# ------------------------------

number_of_students = int(input("Enter number of students: "))

for i in range(number_of_students):

    student_id = input("Student ID: ")
    student_name = input("Student name: ")
    dob = input("Date of birth: ")

    student = [student_id, student_name, dob]

    students.append(student)


# ------------------------------
# Input courses
# ------------------------------

number_of_courses = int(input("\nEnter number of courses: "))

for i in range(number_of_courses):

    course_id = input("Course ID: ")
    course_name = input("Course name: ")
    credit = int(input("Number of credits: "))

    course = [course_id, course_name, credit]

    course_list.append(course)


# ------------------------------
# Input marks
# ------------------------------

for student in students:

    for course in course_list:

        mark = float(input(
            "Enter mark for "
            + student[1]
            + " - "
            + course[1]
            + ": "
        ))

        # Round DOWN to 1 decimal place
        mark = math.floor(mark * 10) / 10

        mark_information = [
            student[0],
            course[0],
            mark
        ]

        marks.append(mark_information)


# ------------------------------
# Calculate GPA
# ------------------------------

def calculate_gpa(student_id):

    weighted_marks = []
    credits = []

    for mark in marks:

        if mark[0] == student_id:

            course_id = mark[1]
            student_mark = mark[2]

            for course in course_list:

                if course[0] == course_id:

                    credit = course[2]

                    weighted_marks.append(
                        student_mark * credit
                    )

                    credits.append(credit)

    weighted_marks = np.array(weighted_marks)
    credits = np.array(credits)

    if np.sum(credits) == 0:
        return 0

    gpa = np.sum(weighted_marks) / np.sum(credits)

    return gpa


# ------------------------------
# Display students
# ------------------------------

def list_students():

    print("\n" + "=" * 40)
    print("STUDENTS")
    print("=" * 40)

    for student in students:

        gpa = calculate_gpa(student[0])

        print(
            "ID:", student[0],
            "| Name:", student[1],
            "| DOB:", student[2],
            "| GPA:", round(gpa, 2)
        )


# ------------------------------
# Display marks
# ------------------------------

def show_marks():

    print("\n" + "=" * 40)
    print("MARKS")
    print("=" * 40)

    for mark in marks:

        print(
            "Student:", mark[0],
            "| Course:", mark[1],
            "| Mark:", mark[2]
        )


# ------------------------------
# Sort students by GPA
# ------------------------------

def sort_students_by_gpa():

    student_gpa = []

    for student in students:

        gpa = calculate_gpa(student[0])

        student_gpa.append([student, gpa])

    student_gpa.sort(
        key=lambda x: x[1],
        reverse=True
    )

    print("\n" + "=" * 40)
    print("STUDENTS SORTED BY GPA")
    print("=" * 40)

    for item in student_gpa:

        student = item[0]
        gpa = item[1]

        print(
            "ID:", student[0],
            "| Name:", student[1],
            "| GPA:", round(gpa, 2)
        )


# ------------------------------
# Display everything
# ------------------------------

list_students()

courses.show_courses(course_list)

show_marks()

sort_students_by_gpa()
```

```python
%%writefile pw4/domains/student.py

class Student:

    def __init__(self, student_id, name, dob):
        self.student_id = student_id
        self.name = name
        self.dob = dob

    def __str__(self):
        return (
            "ID: " + self.student_id
            + " | Name: " + self.name
            + " | DOB: " + self.dob
        )
```

```python
%%writefile pw4/domains/course.py

class Course:

    def __init__(self, course_id, name, credit):
        self.course_id = course_id
        self.name = name
        self.credit = credit

    def __str__(self):
        return (
            "ID: " + self.course_id
            + " | Name: " + self.name
            + " | Credits: " + str(self.credit)
        )
```

```python
%%writefile pw4/domains/mark.py

class Mark:

    def __init__(self, student_id, course_id, mark):
        self.student_id = student_id
        self.course_id = course_id
        self.mark = mark
```

```python
%%writefile pw4/domains/__init__.py

from .student import Student
from .course import Course
from .mark import Mark
```

```python
%%writefile pw4/input.py

from domains import Student, Course, Mark


def input_students():
    students = []

    number_of_students = int(input("Enter number of students: "))

    for i in range(number_of_students):
        student_id = input("Student ID: ")
        name = input("Student name: ")
        dob = input("Date of birth: ")

        student = Student(student_id, name, dob)
        students.append(student)

    return students


def input_courses():
    courses = []

    number_of_courses = int(input("\nEnter number of courses: "))

    for i in range(number_of_courses):
        course_id = input("Course ID: ")
        name = input("Course name: ")
        credit = int(input("Number of credits: "))

        course = Course(course_id, name, credit)
        courses.append(course)

    return courses


def input_marks(students, courses):
    marks = []

    for student in students:
        for course in courses:

            mark = float(input(
                "Enter mark for "
                + student.name
                + " - "
                + course.name
                + ": "
            ))

            mark = int(mark * 10) / 10

            new_mark = Mark(
                student.student_id,
                course.course_id,
                mark
            )

            marks.append(new_mark)

    return marks
```
