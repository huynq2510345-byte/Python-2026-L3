```python
# prime number

n = int(input("Enter number: "))

if n > 1:
    prime = True

    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    if prime:
        print("Your number is a prime number:", n)
    else:
        print("Your number is not a prime number:", n)
else:
    print("Your number is not a prime number:", n)
```

```python
#perfect number
n = int(input("Enter number: "))
sum = 0
for i in range(1, n):
    if n % i == 0:
        sum += i
if sum == n:
    print("Your number is a perfect number:", n)
else:
    print("Your number is not a perfect number:", n)
```

```python
def extract_even(l):
    even = []

    for i in l:
        if i % 2 == 0:
            even.append(i)

    return even


# Test
numbers = [1, 4, 5, 6, 8, 9, 10]
print(extract_even(numbers))
```

```python
#ex9
def factorial(n):
    result = 1
    for i in range(1, n + 1):
      result = result * i
    return result
print(factorial(5))
```

```python
#ex10
def get_divisors(n):
    divisors = []
    for i in range(1, n+1):
        if n % i == 0:
            divisors.append(i)
    return divisors

print(get_divisors(12))
```
